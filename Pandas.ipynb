```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# ==========================================
# STEP 0: Generate Synthetic Dataset (200+ rows)
# ==========================================
n_employees = 250

departments = ["HR", "Sales", "Engineering", "Marketing", "Finance"]

data = {
    "Employee_ID": [f"EMP{i:03d}" for i in range(1, n_employees + 1)],
    "Department": np.random.choice(departments, size=n_employees),
    "Age": np.random.randint(22, 60, size=n_employees),
    "Salary": np.random.randint(30000, 120000, size=n_employees),
    "Experience": np.random.randint(1, 25, size=n_employees),
    "Performance": np.random.randint(40, 100, size=n_employees),
    "Attrition": np.random.choice(
        ["Yes", "No"], size=n_employees, p=[0.25, 0.75]
    ),
}

# Save raw dataset
df_raw = pd.DataFrame(data)
df_raw.to_csv("employee_attrition_raw.csv", index=False)

# ==========================================
# Task 1: Load dataset using Pandas
# ==========================================
df = pd.read_csv("employee_attrition_raw.csv")

# ==========================================
# Task 2: Display first 10 records
# ==========================================
print("--- Task 2: First 10 Records ---")
print(df.head(10))
print()

# ==========================================
# Task 3: Find Average Salary, Age, Experience
# ==========================================
avg_salary = df["Salary"].mean()
avg_age = df["Age"].mean()
avg_exp = df["Experience"].mean()

print("--- Task 3: Averages ---")
print(f"Average Salary:     ₹{avg_salary:,.2f}")
print(f"Average Age:        {avg_age:.2f} years")
print(f"Average Experience: {avg_exp:.2f} years\n")

# ==========================================
# Task 4: Count employees in each department
# ==========================================
dept_counts = df["Department"].value_counts()
print("--- Task 4: Employee Count by Department ---")
print(dept_counts)
print()

# ==========================================
# Task 5: Find department with highest attrition
# ==========================================
attrition_by_dept = (
    df[df["Attrition"] == "Yes"]["Department"].value_counts()
)
highest_attrition_dept = attrition_by_dept.idxmax()
highest_attrition_count = attrition_by_dept.max()

print("--- Task 5: Department with Highest Attrition ---")
print(
    f"{highest_attrition_dept} ({highest_attrition_count} employees left)\n"
)

# ==========================================
# Task 6: Filter employees (Salary > 50000 and Experience > 5 years)
# ==========================================
filtered_df = df[(df["Salary"] > 50000) & (df["Experience"] > 5)]
print(
    f"--- Task 6: Filtered Employees (Found {len(filtered_df)} matches) ---"
)
print(filtered_df.head())
print()

# ==========================================
# Task 7: Sort employees by Salary descending
# ==========================================
sorted_df = df.sort_values(by="Salary", ascending=False)
print("--- Task 7: Top 5 Highest Earners ---")
print(sorted_df[["Employee_ID", "Department", "Salary"]].head())
print()

# ==========================================
# Task 8: Group by Department and calculate Average Salary & Average Performance
# ==========================================
dept_summary = df.groupby("Department").agg(
    {"Salary": "mean", "Performance": "mean"}
)
print("--- Task 8: Department Averages (Salary & Performance) ---")
print(dept_summary)
print()


# ==========================================
# Task 9: Create a new column 'Risk_Level'
# Rules:
# - Attrition = Yes -> High Risk
# - Performance < 60 -> Medium Risk
# - Else -> Low Risk
# ==========================================
def assign_risk(row):
  if row["Attrition"] == "Yes":
    return "High Risk"
  elif row["Performance"] < 60:
    return "Medium Risk"
  else:
    return "Low Risk"


df["Risk_Level"] = df.apply(assign_risk, axis=1)

print("--- Task 9: Risk Level Distribution ---")
print(df["Risk_Level"].value_counts())
print()

# ==========================================
# Final Step: Export cleaned/processed dataset
# ==========================================
df.to_csv("employee_attrition_cleaned.csv", index=False)
print("Cleaned dataset successfully exported to 'employee_attrition_cleaned.csv'")
```
